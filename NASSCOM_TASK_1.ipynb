1.SECOND LARGEST NUMBER

```python
def largest_second_num(lst):
    max = lst[0]
    sl_num = lst[0]
    for num in lst:
        if num > max:
            sl_num = max
            max = num
        elif num > sl_num and num != sl_num:
            sl_num = num
    print(sl_num)
    return lst
```

```python
lst = []
n = int(input())
for i in range(n):
    num = int(input())
    lst.append(num)
largest_second_num(lst)
print(lst)
```

2.FREQUENCY ANALYSIS

```python
def frequency_of_num(lst):
    visited = []
    for i in lst:
        if i not in visited:
            freq = 0
            for j in lst:
                if i == j:
                    freq = freq + 1
            print(i, "-", freq)
            visited.append(i)
```

```python
lst = []
n = int(input())
for i in range(n):
    num = int(input())
    lst.append(num)
frequency_of_num(lst)
```

3.LIST ROTATION

```python
def lst_rotation(lst,k):
    for i in range(k):
        x = lst.pop()
        lst.insert(0,x)
    return lst
```

```python
lst = []
n = int(input())
for i in range(n):
    num = int(input())
    lst.append(num)
k = int(input("k = "))
print(lst_rotation(lst,k))
```

4.REMOVE DUPLICATES

```python
def remove_duplicates(lst):
    removed = []
    for i in lst:
        if i not in removed:
            removed.append(i)
    return removed
```

```python
lst = []
n = int(input())
for i in range(n):
    num = int(input())
    lst.append(num)
output = remove_duplicates(lst)
print(output)
```

5.PAIR SUM

```python
def pair_sum(lst,target):
    ans = []
    for i in range(len(lst)+1):
        for j in range(i+1,len(lst)+1):
            if int(i)+int(j) == target:
                ans += [(i,j)]
    print(ans)
```

```python
lst = []
n = int(input())
for i in range(n):
    num = int(input())
    lst.append(num)
target = int(input("target = "))
pair_sum(lst,target)
```

8. SPIRAL TRAVERSAL

```python
def spiralTraversal(matrix, n):
    top = 0
    bottom = n - 1
    left = 0
    right = n - 1

    while top <= bottom and left <= right:
        for i in range(left, right + 1):
            print(matrix[top][i], end=" ")
        top += 1

        for i in range(top, bottom + 1):
            print(matrix[i][right], end=" ")
        right -= 1

        if top <= bottom:
            for i in range(right, left - 1, -1):
                print(matrix[bottom][i], end=" ")
            bottom -= 1

        if left <= right:
            for i in range(bottom, top - 1, -1):
                print(matrix[i][left], end=" ")
            left += 1
```

```python
n = int(input("Enter size of the matrix: "))

matrix = []

print("Enter matrix elements:")

for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
print(spiralTraversal(matrix,n))
```

10.STOCK BY AND SELL

```python
def stock_buy_sell(prices):
    min_price = prices[0]
    max_profit = 0

    for price in prices:
        if price < min_price:
            min_price = price
        elif price - min_price > max_profit:
            max_profit = price - min_price

    return max_profit
```

```python
n = int(input("Enter value of n: "))
prices = list(map(int, input("Enter stock prices: ").split()))

profit = stock_buy_sell(prices)

print("Maximum Profit:", profit)
```
